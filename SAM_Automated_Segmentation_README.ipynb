## Proje Hakkında

Bu notebook, Meta Research tarafından yayınlanan **Segment Anything Model**’i kullanarak bir görüntüdeki nesneleri herhangi bir etiketleme, eğitim veya prompt ihtiyacı olmadan otomatik olarak segmente eder.

Notebook, görseli okur, SAM ile maskeleri üretir, her maskeyi renkli bir katman olarak görselleştirir ve sonucu çalışma dizinine kaydeder.


## Özellikler

- Görüntü üzerindeki nesneleri otomatik olarak segmente eder.
- Prompt, bounding box veya manuel seçim gerektirmez.
- ViT-B, ViT-L ve ViT-H model seçeneklerini destekler.
- Maskeleri renkli ve yarı şeffaf katmanlarla görselleştirir.
- Segmentasyon sonucunu `.png` formatında kaydeder.
- Colab GPU ortamı için hazırdır.

## Model Seçenekleri

Notebook içinde `MODEL_TYPE` değişkeniyle kullanılacak model seçilebilir.

| Model | Parametre | Yaklaşık Boyut | Kullanım Notu |
|---|---:|---:|---|
| ViT-B | 91M | ~375 MB | Daha hızlı, daha hafif |
| ViT-L | 308M | ~1.2 GB | Dengeli seçenek |
| ViT-H | 636M | ~2.5 GB | En güçlü, en yavaş seçenek |

Varsayılan olarak `vit_h` kullanılır.


````python
# Temel parametreler
IMAGE_NAME = "deneme.jpeg"
MODEL_TYPE = "vit_h"
RESULT_NAME = "segmentasyon_sonucu.png"

# SAM otomatik maske üretici parametreleri
points_per_side = 32
pred_iou_thresh = 0.88
stability_score_thresh = 0.95
min_mask_region_area = 100
````
